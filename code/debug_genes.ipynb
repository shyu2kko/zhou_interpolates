# load mesh

```python
import pyvista as pv

reduced_mesh = joblib.load(projdir + '/data/fsLR-4k_reduced-midthickness_neuromaps.pkl')
```

```python
import pandas as pd

mappings = pd.read_csv(projdir + '/data/beliveau2017_serotonergic_names_mappings.csv')
```

```python
mappings['gene'][0]
```

```python
maptag = mappings['gene'][0]

from interpmodules import deterministic, geospatial, helpers

known_coords = mni_coords.to_numpy()
unknown_coords = reduced_mesh.points

known_vals = expr[maptag]
```

```python
interp_vals = deterministic.interpolate_IDW(X0=known_coords, Y0=known_vals, x1=unknown_coords, power=4)
```

```python
pl = pv.Plotter()

pv.start_xvfb()

pl.add_mesh(reduced_mesh, scalars = interp_vals, cmap = 'magma')
pl.show(jupyter_backend = 'static')
```
